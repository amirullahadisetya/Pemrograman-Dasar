```python
def kategori_usia(usia):
    if usia < 13:
        return "Anak-anak"
    elif usia < 18:
        return "Remaja"
    elif usia < 60:
        return "Dewasa"
    else:
        return "Lansia"

print(kategori_usia(14))   # Output: Dewasa

```

```python
def cari_nama_panjang(daftar_nama):
    nama_panjang = ""
    for nama in daftar_nama:
        if len(nama) > len(nama_panjang):
            nama_panjang = nama
    return nama_panjang


nama_siswa = ["Amir", "Bambang", "Cinta", "Dewi Anjani"]
print("Nama terpanjang adalah:", cari_nama_panjang(nama_siswa))
```

```python
def hitung_stok(barang):
    i = 0
    total_stok = 0
    daftar_barang = list(barang.items())

    while i < len(daftar_barang):
        nama, stok = daftar_barang[i]
        print(f"{nama} tersisa {stok} buah")
        total_stok += stok
        i += 1

    return f"Total seluruh stok barang: {total_stok}"

# Contoh penggunaan
gudang = {"Pensil": 50, "Buku": 30, "Penghapus": 25, "Pulpen": 40}
print(hitung_stok(gudang))
```

```python
def cek_vokal(huruf):
    vokal = ('a', 'e', 'i', 'o', 'u')
    return huruf.lower() in vokal

# Contoh penggunaan
print(cek_vokal('A'))  # Output: True
print(cek_vokal('H'))  # Output: False
```

```python
def hapus_duplikat(data):
    hasil = set()
    for item in data:
        hasil.add(item)
    return hasil

# Contoh penggunaan
angka = [1, 2, 2, 3, 3, 4, 5, 5]
print("Data tanpa duplikat:", hapus_duplikat(angka))
```
