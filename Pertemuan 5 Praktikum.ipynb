```python
print("--- Soal 2 ---")
kode_berhenti = 4232401016
input_pengguna = 0 # Inisialisasi awal

while input_pengguna < kode_berhenti:
    try:
        input_pengguna = int(input("Masukkan angka (4232401016 untuk berhenti): "))
    except ValueError:
        print("Input tidak valid. Harap masukkan angka.")

print("Perulangan telah dihentikan oleh sentinel (4232401016).")

```

```python
print("--- Soal 3 ---")
harga_barang = [15000, 25000, 10000, 5000, 13000, 20000, 22000, 30000, 80000, 18000]
total_belanja = 0

for harga in harga_barang:
    total_belanja += harga

print("Total biaya belanja adalah:", total_belanja)

```

```python
print("--- Soal 4 ---")
percobaan_maksimal = 3
percobaan_saat_ini = 1

while percobaan_saat_ini <= percobaan_maksimal:
    print(f"Percobaan ke-{percobaan_saat_ini}...")
    # Asumsi ada logika input password di sini
    percobaan_saat_ini += 1

if percobaan_saat_ini == percobaan_maksimal + 1:
    print("Kesempatan habis.")
else:
    print("Akses diterima.")
```

```python
print("--- Soal 5 ---")
nama_proyek = "Politeknik Negeri Batam"
hasil_output = ""

for karakter in nama_proyek:
    hasil_output += karakter + "-"

print(hasil_output) # Output: P-o-l-i-t-e-k-n-i-k- -N-e-g-e-r-i- -B-a-t-a-m-
```

```python
print("--- Soal 6 ---")
# Countdown dari 5 ke 1
for i in range(255, 0, -1):
    print(i)
```

```python
print("--- Soal 7 ---")
kelurahan = ["Muka Kuning", "Baloi Permai", "Bengkong Indah", "Tembesi", "Sei Panas", "Kabil", "Teluk Tering", "Nongsa", "Kampung Seraya", "Buliang", "Sungai Langkai", "Tanjung Sengkuang", "Kibing", "Tanjung Riau"]

for i, nama in enumerate(kelurahan):
    # i adalah indeks, nama adalah elemen
    print(f"kelurahan ke-{i + 1}: {nama}")
```

```python
print("--- Soal 8 ---")
data_angka = [10, 20, 30, 1, 0, 50, 60]

for angka in data_angka:
    if angka < 1:
        print(f"angka negatif ditemukan: {angka}")
        break
    print(angka)

print("Program berhenti karena angka 0 ditemukan.")
```

```python
print("--- Soal 9 ---")
total = 0
bilangan = 1
batas = 20

while total <= batas:
    total += bilangan
    bilangan += 1

print("Total terakhir:", total)
```
